```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
data = pd.read_csv('Bakery Sales.csv')
data['datetime'] = pd.to_datetime(data['datetime'])  # Convert to datetime
data.set_index('datetime', inplace=True)  # Set datetime as index

# Resample by day and sum total sales
daily_sales = data['total'].resample('D').sum()

# Plot
plt.figure(figsize=(12, 6))
daily_sales.plot(color='blue', linewidth=2)
plt.title('Daily Total Sales Trend (2019)', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Total Sales (₩)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Group by day of week and calculate mean sales
avg_sales_by_day = data.groupby('day of week')['total'].mean().sort_values()

# Plot
plt.figure(figsize=(10, 6))
avg_sales_by_day.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Sales by Day of Week', fontsize=16)
plt.xlabel('Day of Week', fontsize=12)
plt.ylabel('Average Sales (₩)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(data['total'], bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
plt.title('Distribution of Total Sales per Transaction', fontsize=16)
plt.xlabel('Total Sales (₩)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(data['americano'], data['croissant'], color='coral', alpha=0.6, edgecolor='black')
plt.title('Relationship Between Americano and Croissant Sales', fontsize=16)
plt.xlabel('Americano Sales (Count)', fontsize=12)
plt.ylabel('Croissant Sales (Count)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
